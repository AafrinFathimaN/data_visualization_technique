```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/91754/Downloads/transactions.csv")

df["Timestamp"] = pd.to_datetime(df["Timestamp"])

daily_amount = df.groupby(df["Timestamp"].dt.date)["Amount (INR)"].sum()

plt.plot(daily_amount.index, daily_amount.values)

plt.title("Daily Transaction Amount")
plt.xlabel("Date")
plt.ylabel("Total Transaction Amount (INR)")

plt.xticks(rotation=45)
plt.show()
```

```python
status_count = df["Status"].value_counts()

plt.bar(status_count.index, status_count.values)

plt.title("Transaction Status Comparison")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
plt.pie(
    status_count,
    labels=status_count.index,
    autopct="%1.1f%%"
)

plt.title("Transaction Status Distribution")

plt.show()
```

```python
amount = df["Amount (INR)"]

plt.hist(amount, bins=5)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount (INR)")
plt.ylabel("Frequency")

plt.show()
```

```python
transaction_number = range(1, len(df) + 1)
amount = df["Amount (INR)"]

plt.scatter(transaction_number, amount)

plt.title("Transaction Number vs Transaction Amount")
plt.xlabel("Transaction Number")
plt.ylabel("Transaction Amount (INR)")

plt.show()
```

```python
plt.boxplot(amount)

plt.title("Transaction Amount Distribution")
plt.ylabel("Transaction Amount (INR)")

plt.show()
```
