```python
import pandas as pd

# Load dataset
df = pd.read_csv("C:/Users/91754/Desktop/dvt/consolidated_coin_data.csv")
# Create a copy for processing
crypto = df.copy()

# Remove duplicate records
crypto.drop_duplicates(inplace=True)

# Convert date to datetime
crypto["Date"] = pd.to_datetime(crypto["Date"], errors="coerce")

# Convert numerical columns
numeric_columns = [
    "Open",
    "High",
    "Low",
    "Close",
    "Volume",
    "Market Cap"
]

for col in numeric_columns:
    crypto[col] = (
        crypto[col]
        .astype(str)
        .str.replace(",", "", regex=False)
    )
    crypto[col] = pd.to_numeric(crypto[col], errors="coerce")

# Handle missing values
for col in numeric_columns:
    crypto[col] = crypto[col].fillna(crypto[col].median())

# Create daily return
crypto["daily_return"] = (
    (crypto["Close"] - crypto["Open"]) /
    crypto["Open"]
) * 100

# Create price range
crypto["price_range"] = crypto["High"] - crypto["Low"]

# Create volatility
crypto["volatility"] = (
    (crypto["High"] - crypto["Low"]) /
    crypto["Open"]
) * 100

# Replace any remaining missing calculated values
crypto["daily_return"] = crypto["daily_return"].fillna(0)
crypto["price_range"] = crypto["price_range"].fillna(0)
crypto["volatility"] = crypto["volatility"].fillna(0)

# Standardize column names
crypto.columns = (
    crypto.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

print("Original Shape:", df.shape)
print("Cleaned Shape:", crypto.shape)

crypto.head()

print(crypto.isna().sum())
```

```python
crypto.to_csv("crypto_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
